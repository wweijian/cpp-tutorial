# C++ for C Programmers: Familiar Ground

If you know C, you already know a large portion of C++. The two languages share the same basic syntax for operators, control flow, functions, pointers, and arrays. In this notebook we will confirm that your existing C knowledge carries over directly, while also pointing out the subtle differences and a few traps to watch for.

By the end of this notebook you will feel confident that C++ is an extension of C, not a replacement that requires starting from scratch.

## Basic Syntax

Arithmetic, comparison, logical, and assignment operators are identical in C and C++.

| Category | Operators |
|---|---|
| Arithmetic | `+`, `-`, `*`, `/`, `%`, `++`, `--` |
| Comparison | `==`, `!=`, `<`, `>`, `<=`, `>=` |
| Logical | `&&`, `||`, `!` |
| Bitwise | `&`, `|`, `^`, `~`, `<<`, `>>` |
| Assignment | `=`, `+=`, `-=`, `*=`, `/=`, `%=` |

All of these work exactly as in C.

````c++
#include <iostream>

int a = 10;
int b = 3;

// Arithmetic
std::cout << "10 + 3 = " << (a + b) << std::endl;
std::cout << "10 - 3 = " << (a - b) << std::endl;
std::cout << "10 * 3 = " << (a * b) << std::endl;
std::cout << "10 / 3 = " << (a / b) << std::endl;  // integer division
std::cout << "10 % 3 = " << (a % b) << std::endl;

// Comparison (result is bool in C++, but prints as 0/1)
std::cout << "10 > 3 ?  " << (a > b)  << std::endl;
std::cout << "10 == 3 ? " << (a == b) << std::endl;

// Logical
bool flag1 = (a > 5) && (b < 10);
bool flag2 = (a < 5) || (b < 10);
std::cout << "(a>5 && b<10): " << flag1 << std::endl;
std::cout << "(a<5 || b<10): " << flag2 << std::endl;
// Expected output:
// 10 + 3 = 13
// 10 - 3 = 7
// 10 * 3 = 30
// 10 / 3 = 3
// 10 % 3 = 1
// 10 > 3 ?  1
// 10 == 3 ? 0
// (a>5 && b<10): 1
// (a<5 || b<10): 1
````

## Primitive Types

All C primitive types are available in C++:

| Type | Typical size | Notes |
|---|---|---|
| `char` | 1 byte | signed or unsigned depending on platform |
| `int` | 4 bytes | signed integer |
| `unsigned int` | 4 bytes | non-negative integers |
| `long` | 4 or 8 bytes | platform-dependent |
| `float` | 4 bytes | single-precision floating point |
| `double` | 8 bytes | double-precision floating point |
| `bool` | 1 byte | **C++ keyword** — `true` or `false` |

**Key difference:** In C, `bool` requires `#include <stdbool.h>`. In C++, `bool`, `true`, and `false` are built-in keywords — no header needed.

````c++
#include <iostream>

// sizeof() works exactly as in C
std::cout << "sizeof(char)   = " << sizeof(char)   << " byte(s)" << std::endl;
std::cout << "sizeof(int)    = " << sizeof(int)    << " byte(s)" << std::endl;
std::cout << "sizeof(long)   = " << sizeof(long)   << " byte(s)" << std::endl;
std::cout << "sizeof(float)  = " << sizeof(float)  << " byte(s)" << std::endl;
std::cout << "sizeof(double) = " << sizeof(double) << " byte(s)" << std::endl;
std::cout << "sizeof(bool)   = " << sizeof(bool)   << " byte(s)" << std::endl;

// bool is a first-class type in C++
bool isReady = true;
bool isDone  = false;
std::cout << "isReady = " << isReady << std::endl;  // prints 1
std::cout << "isDone  = " << isDone  << std::endl;  // prints 0

// Use std::boolalpha to print true/false as words
std::cout << std::boolalpha;
std::cout << "isReady = " << isReady << std::endl;  // prints true
std::cout << "isDone  = " << isDone  << std::endl;  // prints false
````

## Control Flow

`if`/`else`, `while`, `do-while`, and `for` loops work exactly as in C. There is nothing new to learn here — just confirm your C knowledge applies.

````c++
#include <iostream>

// if / else if / else
int score = 72;
if (score >= 90) {
    std::cout << "Grade: A" << std::endl;
} else if (score >= 75) {
    std::cout << "Grade: B" << std::endl;
} else if (score >= 60) {
    std::cout << "Grade: C" << std::endl;
} else {
    std::cout << "Grade: F" << std::endl;
}

// for loop
std::cout << "Counting: ";
for (int i = 0; i < 5; i++) {
    std::cout << i << " ";
}
std::cout << std::endl;

// while loop
int n = 1;
while (n < 16) {
    n *= 2;
}
std::cout << "First power of 2 >= 16: " << n << std::endl;

// do-while loop (runs at least once)
int counter = 0;
do {
    counter++;
} while (counter < 3);
std::cout << "counter = " << counter << std::endl;
// Expected output:
// Grade: C
// Counting: 0 1 2 3 4
// First power of 2 >= 16: 16
// counter = 3
````

## The switch Statement

The `switch` statement works identically to C. It evaluates an integer expression and jumps to the matching `case`. Without a `break`, execution **falls through** to the next case.

### Fall-through: intentional vs accidental
Fall-through can be useful (e.g., multiple cases doing the same thing), but forgetting `break` is one of the most common C/C++ bugs. **Always add a `break` unless fall-through is intentional, and add a comment when it is.**

````c++
#include <iostream>

char grade = 'B';

switch (grade) {
    case 'A':
        std::cout << "Excellent!" << std::endl;
        break;
    case 'B':
        std::cout << "Good job." << std::endl;
        break;
    case 'C':
        std::cout << "Passing." << std::endl;
        break;
    case 'D':
        // intentional fall-through: D and F both mean failing
    case 'F':
        std::cout << "Failing." << std::endl;
        break;
    default:
        std::cout << "Unknown grade." << std::endl;
        break;
}
// Expected output: Good job.
````

````c++
#include <iostream>

// Demonstration of accidental fall-through (a common bug)
int x = 2;
switch (x) {
    case 1:
        std::cout << "one" << std::endl;
    case 2:
        std::cout << "two" << std::endl;  // missing break!
    case 3:
        std::cout << "three" << std::endl;  // this runs too!
        break;
    default:
        std::cout << "other" << std::endl;
}
// Expected output:
// two
// three    <-- fall-through! case 3 ran even though x==2
````

## Functions

Function syntax is identical to C: return type, name, parameter list, body. C++ adds some extras (default parameters, overloading, references) that we will cover in later notebooks. For now, confirm the basics work the same.

````c++
#include <iostream>

// Function declaration (prototype) — same as C
int add(int x, int y);
double power(double base, int exp);

// Function definitions
int add(int x, int y) {
    return x + y;
}

double power(double base, int exp) {
    double result = 1.0;
    for (int i = 0; i < exp; i++) {
        result *= base;
    }
    return result;
}

// Function calls
std::cout << "add(3, 4) = "   << add(3, 4)    << std::endl;
std::cout << "2^10 = "        << power(2, 10)  << std::endl;
// Expected output:
// add(3, 4) = 7
// 2^10 = 1024
````

## Pointers (Review)

Pointers in C++ work exactly as in C. `*` dereferences a pointer, `&` takes the address of a variable. We will cover **references** (a C++-only feature that is often cleaner than pointers) in a later notebook.

````c++
#include <iostream>

int value = 42;
int *ptr = &value;  // ptr holds the address of value

std::cout << "value      = " << value  << std::endl;
std::cout << "&value     = " << &value << std::endl;  // address
std::cout << "ptr        = " << ptr    << std::endl;  // same address
std::cout << "*ptr       = " << *ptr   << std::endl;  // dereference: 42

// Modify through pointer
*ptr = 100;
std::cout << "After *ptr = 100: value = " << value << std::endl;

// Pointer to pointer
int **pptr = &ptr;
std::cout << "**pptr = " << **pptr << std::endl;
// Expected output:
// value      = 42
// &value     = 0x...  (some address)
// ptr        = 0x...  (same address)
// *ptr       = 42
// After *ptr = 100: value = 100
// **pptr = 100
````

## Arrays

Fixed-size arrays and array iteration are identical to C. The same caveats apply: no bounds checking, array name decays to a pointer, `sizeof` on an array gives total bytes not element count.

````c++
#include <iostream>

int scores[5] = {95, 82, 73, 88, 91};
int length = sizeof(scores) / sizeof(scores[0]);  // 5

std::cout << "Scores: ";
for (int i = 0; i < length; i++) {
    std::cout << scores[i];
    if (i < length - 1) std::cout << ", ";
}
std::cout << std::endl;

// Sum using a pointer (just like C)
int sum = 0;
int *p = scores;
for (int i = 0; i < length; i++) {
    sum += *(p + i);
}
std::cout << "Sum = " << sum << std::endl;
std::cout << "Average = " << (double)sum / length << std::endl;
// Expected output:
// Scores: 95, 82, 73, 88, 91
// Sum = 429
// Average = 85.8
````

## Macros

The C preprocessor is available in C++ without changes. `#define`, `#ifdef`, `#ifndef`, `#include`, etc. all work the same.

### Constants with #define
```cpp
#define PI 3.14159
#define MAX_SIZE 100
```

### Function-like macros
```cpp
#define SQUARE(x) ((x) * (x))
```
Note the parentheses around `x` — without them, `SQUARE(1+2)` expands to `1+2 * 1+2 = 5` instead of `9`.

### Header guards
Use `#ifndef` to prevent a header from being included more than once:
```cpp
// MyClass.hpp
#ifndef MYCLASS_HPP
#define MYCLASS_HPP

class MyClass {
    // ...
};

#endif  // MYCLASS_HPP
```
This is not runnable in a notebook (it is a `.hpp` pattern), but you will use it in every header file you write.

### Macros: limitations
- **No type safety** — `SQUARE` works on any type but cannot enforce types.
- **No scope** — macros are global; they can clash with variable names.
- **Hard to debug** — errors point to the expanded code, not the macro definition.
- **Double evaluation** — `SQUARE(i++)` expands to `((i++) * (i++))` — undefined behaviour!

C++ provides better alternatives: `const` variables (for constants) and `inline` functions (for function-like macros). More on this in the Modern C++ section below.

````c++
#include <iostream>

#define PI       3.14159
#define MAX_SIZE 100
#define SQUARE(x) ((x) * (x))   // parentheses around x are critical!

double radius = 5.0;
double area = PI * SQUARE(radius);
std::cout << "Area of circle (r=5): " << area << std::endl;

// Pitfall: without inner parentheses, SQUARE(1+2) would be wrong
std::cout << "SQUARE(1+2) = " << SQUARE(1+2) << std::endl;  // correct: 9

std::cout << "MAX_SIZE = " << MAX_SIZE << std::endl;
// Expected output:
// Area of circle (r=5): 78.5397
// SQUARE(1+2) = 9
// MAX_SIZE = 100
````

## const in C++

`const` is significantly more powerful in C++ than in C:

- In C, a `const` variable at file scope is still `extern` by default — it leaks into other translation units.
- In **C++98**, a `const` variable at namespace scope has **internal linkage by default** — it is private to the translation unit. This makes `const int MAX = 100;` in a header file safe.
- `const` can be applied to pointers in two ways: pointer-to-const vs const-pointer.
- `const` member functions promise not to modify the object (covered in the classes notebook).

**Prefer `const` over `#define` for named constants.** It has a type, a scope, and plays nicely with the debugger.

````c++
#include <iostream>

// const variable (prefer this over #define for constants)
const int MAX_PLAYERS = 4;
const double GRAVITY  = 9.81;

std::cout << "Max players: " << MAX_PLAYERS << std::endl;
std::cout << "Gravity: "     << GRAVITY     << std::endl;

// Pointer to const: cannot modify the value through the pointer
int val = 42;
const int *pConst = &val;   // pointer to const int
// *pConst = 10;            // ERROR: cannot modify through pConst
val = 10;                   // OK: val itself is not const
std::cout << "val = " << val << ", *pConst = " << *pConst << std::endl;

// Const pointer: the pointer itself cannot be reassigned
int a = 1, b = 2;
int * const constPtr = &a;  // const pointer to int
*constPtr = 99;             // OK: can modify the value
// constPtr = &b;           // ERROR: cannot reassign the pointer
std::cout << "a = " << a << std::endl;
// Expected output:
// Max players: 4
// Gravity: 9.81
// val = 10, *pConst = 10
// a = 99
````

## Modern C++ (C++11 and Beyond)

### nullptr vs NULL vs 0

In C and C++98, the null pointer is represented as `0` or the macro `NULL` (which expands to `0` or `(void*)0`). This causes ambiguity:

````c++
#include <iostream>

// Ambiguity with NULL / 0
void process(int n) {
    std::cout << "process(int): " << n << std::endl;
}
void process(int *p) {
    std::cout << "process(int*): " << (p ? "non-null" : "null") << std::endl;
}

process(0);        // calls process(int) — might not be what you wanted
process(NULL);     // also calls process(int) — ambiguous!
process(nullptr);  // C++11: unambiguously calls process(int*)
// Expected output:
// process(int): 0
// process(int): 0
// process(int*): null
````

### constexpr — a better #define

C++11 introduces `constexpr` for compile-time constants. Unlike `#define`, `constexpr` has a type, respects scope, and works with the debugger:

```cpp
// C++98
#define MAX_SIZE 100      // no type, no scope
const int maxSize = 100;  // has type, but not guaranteed compile-time

// C++11
constexpr int MAX_SIZE = 100;     // guaranteed compile-time, has type
constexpr double PI = 3.14159265; // works for floating point too
```

### inline functions — a better function-like macro

Already available in C++98 but worth emphasising: prefer `inline` functions over function-like macros. They have types, scope, and no double-evaluation problem:

```cpp
// Macro (dangerous):
#define SQUARE(x) ((x) * (x))

// Inline function (safe):
inline int square(int x) { return x * x; }
// square(i++) increments i once, as expected
```
