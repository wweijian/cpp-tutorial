# Introduction to Jupyter Notebooks

Welcome to the C++ tutorial series! Before we dive into code, we need to get comfortable with the tool we will be using throughout this course: **Jupyter Notebooks**.

Jupyter Notebooks let you write and run code interactively, mixing executable cells with formatted text, explanations, and exercises — all in one document. For a programming course, this is ideal: you can read the explanation and immediately experiment with the code right below it.

In this notebook we will cover the Jupyter interface, how to run cells, useful keyboard shortcuts, and how the C++ kernel works.

## The Notebook Interface

When you open a notebook you will see several key areas:

- **Title bar** — shows the name of the notebook file (e.g., `00_jupyter_intro.ipynb`).
- **Menu bar** — File, Edit, View, Insert, Cell, Kernel, Help. Most operations are here.
- **Toolbar** — icon shortcuts for common actions: save, add cell, cut/copy/paste, run, stop kernel, restart kernel, cell type selector.
- **Cell area** — the main working area. A notebook is a sequence of cells stacked vertically.
- **Cell type selector** — dropdown in the toolbar that shows `Code`, `Markdown`, or `Raw`. Use this to change a cell's type.
- **Execution counter** — to the left of each code cell, shown as `In [N]:`. A `*` means the cell is currently running.
- **Status bar** — at the bottom, shows the kernel status (Idle / Busy) and the current mode (Command / Edit).

There are two **modes**:
- **Edit mode** (green border) — you are typing inside a cell. Press `Escape` to leave.
- **Command mode** (blue border) — you are navigating between cells. Press `Enter` to enter edit mode.

## Cell Types

There are two main cell types you will use:

### Markdown Cells
Markdown cells contain formatted text. They support headings (`# H1`, `## H2`), **bold**, *italic*, bullet lists, numbered lists, code spans (`` `code` ``), and more. This cell is a markdown cell. Double-click it to see its raw source.

### Code Cells
Code cells contain executable code. In this course the kernel is **xcpp11** (xeus-cling), which runs C++. When you run a code cell, the output appears directly below it.

The cell below is a code cell. Run it with `Shift+Enter`.

```c++
#include <iostream>

std::cout << "I am a code cell!" << std::endl;
// Expected output: I am a code cell!
```

## Running Cells

There are three main ways to run a cell:

| Shortcut | Action |
|---|---|
| `Shift + Enter` | Run cell and move to the next cell (creates one if at the end) |
| `Ctrl + Enter` | Run cell and stay on the same cell |
| `Alt + Enter` | Run cell and insert a new empty cell below |

You can also click the **Run** button (▶) in the toolbar.

Try running the arithmetic cell below with each shortcut to feel the difference.

```c++
#include <iostream>

int a = 7;
int b = 3;

std::cout << "a + b = " << (a + b) << std::endl;
std::cout << "a - b = " << (a - b) << std::endl;
std::cout << "a * b = " << (a * b) << std::endl;
std::cout << "a / b = " << (a / b) << std::endl;   // integer division!
std::cout << "a % b = " << (a % b) << std::endl;
// Expected output:
// a + b = 10
// a - b = 4
// a * b = 21
// a / b = 2
// a % b = 1
```

## Keyboard Shortcuts

Learning shortcuts will make you much faster. The shortcuts below work in **Command mode** (press `Escape` first if you are in edit mode).

| Shortcut | Action |
|---|---|
| `A` | Insert a new cell **above** the current cell |
| `B` | Insert a new cell **below** the current cell |
| `D, D` (press D twice) | **Delete** the current cell |
| `M` | Change cell type to **Markdown** |
| `Y` | Change cell type to **Code** |
| `Ctrl + Z` | **Undo** the last cell deletion |
| `Enter` | Enter **edit mode** for the selected cell |
| `Escape` | Return to **command mode** |
| `Shift + M` | **Merge** selected cells |
| `Ctrl + Shift + -` | **Split** cell at cursor |
| `H` | Show all keyboard shortcuts (help popup) |

**Tip:** You can also view all shortcuts via **Help > Keyboard Shortcuts** in the menu bar.

## The Kernel

The **kernel** is the engine that executes your code. In standard Python notebooks the kernel is IPython. In this course we use **xeus-cling**, a C++ kernel that supports C++11 syntax.

Key things to know about the kernel:

- The kernel starts automatically when you open a notebook.
- The kernel status is shown in the top-right corner: **Idle** (ready) or **Busy** (running).
- If your code hangs (infinite loop, etc.) you can **interrupt** the kernel: `Kernel > Interrupt` or press the ■ stop button.
- To reset everything and start fresh: **`Kernel > Restart & Run All`**. This clears all variables and re-runs every cell from top to bottom.
- **`Kernel > Restart`** clears all variables but does NOT re-run cells.

### What xeus-cling does differently from a normal C++ compiler

- No `int main()` wrapper needed — you write statements directly.
- `#include` directives work as expected.
- You can define functions and classes in cells.
- State (variables, functions, types) persists across cells in the same session.

## Cell State Persistence

This is one of the most important things to understand about Jupyter Notebooks: **all cells share the same state within a session**.

If you define a variable in one cell, you can use it in any later cell — as long as you ran the first cell first.

Run the two cells below in order to see this in action.

```c++
// Cell 1: Define a variable
int x = 10;
std::cout << "x has been defined as " << x << std::endl;
// Expected output: x has been defined as 10
```

```c++
#include <iostream>

// Cell 2: Use the variable defined in Cell 1
// x is still available because cells share state!
std::cout << "x times 2 = " << (x * 2) << std::endl;
// Expected output: x times 2 = 20
```

**Warning:** If you restart the kernel, `x` will be undefined and Cell 2 will fail. Always run cells from top to bottom, or use **Kernel > Restart & Run All** to ensure a clean, reproducible state.

Also, **order matters**. If you run Cell 2 before Cell 1, you will get an error because `x` has not been defined yet.

## Saving Your Work

Jupyter has several ways to save your notebook:

- **`Ctrl + S`** — saves the notebook immediately. Do this often.
- **File > Save and Checkpoint** — saves AND creates a named checkpoint you can revert to.
- **Autosave** — Jupyter autosaves every 2 minutes by default. You can see "Last Checkpoint" in the title area.
- **File > Download as > Notebook (.ipynb)** — download a copy of the file.

Your notebook file is stored on disk as a `.ipynb` file, which is actually JSON. It stores the source of every cell AND the outputs from the last run.

**Good habit:** Before submitting or sharing a notebook, use **Kernel > Restart & Run All** to make sure everything runs cleanly from scratch, then save.
