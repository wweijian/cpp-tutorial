# 04 — iostream and Strings

C++ replaces `printf`/`scanf` with a cleaner stream-based I/O system and a proper string type. Instead of format strings and type codes like `%d` or `%s`, you use the `<<` and `>>` operators with stream objects. The standard library also provides `std::string`, a real class that manages its own memory — no more fixed-size `char` arrays and `strcpy`.


## Output with cout

`std::cout` is the standard output stream. You send data to it with the insertion operator `<<`. You can chain multiple `<<` in one statement.

**`std::endl` vs `"\n"`:** Both end a line, but `std::endl` also *flushes* the output buffer (forces the data to be written immediately). Use `"\n"` for performance in tight loops; use `std::endl` when you need guaranteed immediate output (e.g., before a crash or blocking operation).


```c++
#include <iostream>

int score = 42;
double pi = 3.14159;
std::string language = "C++";

std::cout << "Score: " << score << std::endl;
std::cout << "Pi: " << pi << std::endl;
std::cout << "Language: " << language << "\n";

// Chaining: print multiple values on one line
std::cout << "Score=" << score << ", Pi=" << pi << ", Lang=" << language << std::endl;
// Expected output:
// Score: 42
// Pi: 3.14159
// Language: C++
// Score=42, Pi=3.14159, Lang=C++
```

```c++
#include <iostream>

// Printing different numeric types
char ch = 'A';
unsigned int u = 4294967295U;
long bigNum = 1000000L;
bool flag = true;

std::cout << "char: " << ch << std::endl;
std::cout << "unsigned int max: " << u << std::endl;
std::cout << "long: " << bigNum << std::endl;
std::cout << "bool: " << flag << std::endl;  // prints 1 (true)
std::cout << std::boolalpha << "bool: " << flag << std::endl;  // prints true
// Expected output:
// char: A
// unsigned int max: 4294967295
// long: 1000000
// bool: 1
// bool: true
```

## Error Output: cerr and clog

C++ provides two additional output streams for error and log messages:

- **`std::cerr`** — unbuffered: output is written immediately. Use for error messages so they appear even if the program crashes before flushing.
- **`std::clog`** — buffered: more efficient for high-volume logging. May not appear immediately.

Both write to `stderr` (not `stdout`), so they can be redirected separately in the shell.


```c++
#include <iostream>

int divisor = 0;

if (divisor == 0) {
    std::cerr << "Error: division by zero!" << std::endl;  // immediate
} else {
    std::cout << 100 / divisor << std::endl;
}

// clog is good for diagnostic messages that don't need to be immediate
std::clog << "[LOG] Program reached checkpoint 1" << std::endl;
std::cout << "Program continues..." << std::endl;
// Expected output:
// Error: division by zero!
// [LOG] Program reached checkpoint 1
// Program continues...
```

## Stream Formatting

The `<iomanip>` header provides manipulators to control output format:

| Manipulator | Effect |
|---|---|
| `std::setw(n)` | Set minimum field width (applies to next output only) |
| `std::setfill(c)` | Set fill character (default is space) |
| `std::left` / `std::right` | Left/right align within field |
| `std::fixed` | Use fixed-point notation for floats |
| `std::setprecision(n)` | Set decimal places (with `fixed`) or significant digits |


```c++
#include <iostream>
#include <iomanip>

// Formatted table output
std::cout << std::left
          << std::setw(15) << "Name"
          << std::setw(8)  << "Score"
          << std::setw(10) << "Grade"
          << std::endl;

std::cout << std::setfill('-') << std::setw(33) << "" << std::setfill(' ') << std::endl;

std::cout << std::left
          << std::setw(15) << "Alice"
          << std::setw(8)  << 95
          << std::setw(10) << "A"
          << std::endl;

std::cout << std::left
          << std::setw(15) << "Bob"
          << std::setw(8)  << 78
          << std::setw(10) << "C+"
          << std::endl;

std::cout << std::left
          << std::setw(15) << "Charlie"
          << std::setw(8)  << 88
          << std::setw(10) << "B+"
          << std::endl;
// Expected output:
// Name           Score   Grade
// ---------------------------------
// Alice          95      A
// Bob            78      C+
// Charlie        88      B+
```

```c++
#include <iostream>
#include <iomanip>

double price = 9.5;
double tax = 0.19;
double total = price * (1.0 + tax);

std::cout << std::fixed << std::setprecision(2);
std::cout << "Price:  $" << std::setw(8) << price << std::endl;
std::cout << "Tax:    $" << std::setw(8) << price * tax << std::endl;
std::cout << "Total:  $" << std::setw(8) << total << std::endl;
// Expected output:
// Price:  $    9.50
// Tax:    $    1.81
// Total:  $   11.32
```

## std::string Basics

`std::string` is a class from `<string>`. It is **not** a `char*` — it manages its own memory, knows its own length, and can grow dynamically. You never need to worry about buffer overflows.


```c++
#include <iostream>
#include <string>

// Declaration and initialization
std::string name = "Alice";
std::string greeting("Hello, world!");
std::string empty;  // default-initialized to empty string

std::cout << "name: " << name << std::endl;
std::cout << "greeting: " << greeting << std::endl;
std::cout << "empty: '" << empty << "'" << std::endl;

// Length and size (they are identical)
std::cout << "Length of name: " << name.length() << std::endl;
std::cout << "Size of greeting: " << greeting.size() << std::endl;

// Empty check
std::cout << "Is empty string empty? " << (empty.empty() ? "yes" : "no") << std::endl;
std::cout << "Is name empty? " << (name.empty() ? "yes" : "no") << std::endl;
// Expected output:
// name: Alice
// greeting: Hello, world!
// empty: ''
// Length of name: 5
// Size of greeting: 13
// Is empty string empty? yes
// Is name empty? no
```

```c++
#include <iostream>
#include <string>

// Accessing individual characters
std::string word = "hello";
std::cout << "First char: " << word[0] << std::endl;
std::cout << "Last char:  " << word[word.length() - 1] << std::endl;

// Modifying characters
word[0] = 'H';
std::cout << "After modify: " << word << std::endl;
// Expected output:
// First char: h
// Last char:  o
// After modify: Hello
```

## String Operations

`std::string` supports rich operations: concatenation, comparison, searching, substitution.


```c++
#include <iostream>
#include <string>

std::string first = "Hello";
std::string second = " World";

// Concatenation with + (creates a new string)
std::string combined = first + second;
std::cout << "Concatenated: " << combined << std::endl;

// += appends in place
first += "!";
std::cout << "After +=: " << first << std::endl;

// append() — same as +=
second.append(" :-)");
std::cout << "After append: " << second << std::endl;
// Expected output:
// Concatenated: Hello World
// After +=: Hello!
// After append:  World :-)
```

```c++
#include <iostream>
#include <string>

std::string a = "apple";
std::string b = "banana";
std::string c = "apple";

// Comparison operators (lexicographic)
std::cout << "a == c: " << (a == c ? "true" : "false") << std::endl;
std::cout << "a == b: " << (a == b ? "true" : "false") << std::endl;
std::cout << "a != b: " << (a != b ? "true" : "false") << std::endl;
std::cout << "a < b:  " << (a < b  ? "true" : "false") << std::endl;  // 'a' < 'b'
std::cout << "b > a:  " << (b > a  ? "true" : "false") << std::endl;
// Expected output:
// a == c: true
// a == b: false
// a != b: true
// a < b:  true
// b > a:  true
```

```c++
#include <iostream>
#include <string>

std::string sentence = "the quick brown fox";

// substr(pos, len) — extract a substring
std::string sub = sentence.substr(4, 5);  // start at index 4, length 5
std::cout << "substr(4,5): " << sub << std::endl;

// find(str) — returns position of first match, or string::npos if not found
size_t pos = sentence.find("brown");
if (pos != std::string::npos) {
    std::cout << "Found 'brown' at index: " << pos << std::endl;
} else {
    std::cout << "'brown' not found" << std::endl;
}

size_t notFound = sentence.find("elephant");
std::cout << "'elephant' found: " << (notFound != std::string::npos ? "yes" : "no") << std::endl;

// replace(pos, len, str) — replace len chars starting at pos with str
std::string s2 = "the quick brown fox";
s2.replace(10, 5, "red");  // replace "brown" with "red"
std::cout << "After replace: " << s2 << std::endl;
// Expected output:
// substr(4,5): quick
// Found 'brown' at index: 10
// 'elephant' found: no
// After replace: the quick red fox
```

## Reading Full Lines

`std::cin >>` reads one token (stops at whitespace). To read an entire line including spaces, use `std::getline(std::cin, str)`.

**The `cin.ignore()` trap:** After `cin >> someVar`, a newline character `'\n'` remains in the input buffer. The next `getline()` call would immediately read that empty line. Fix: call `std::cin.ignore()` to discard that leftover newline.


```c++
#include <iostream>
#include <string>
#include <sstream>

// Simulating getline behavior with a string stream (since cin is tricky in notebooks)
std::string inputData = "42\nHello World from C++";
std::istringstream iss(inputData);

int number;
iss >> number;

// After reading the int, the newline is still in the buffer
// We must ignore it before calling getline
iss.ignore();  // same pattern as: std::cin.ignore();

std::string fullLine;
std::getline(iss, fullLine);

std::cout << "Number: " << number << std::endl;
std::cout << "Full line: " << fullLine << std::endl;
// Expected output:
// Number: 42
// Full line: Hello World from C++
```

## C-string Interop

Sometimes you need to interface with C libraries that expect `char*`. `std::string` provides `c_str()` which returns a `const char*` pointing to the internal buffer. You can also construct a `std::string` from a `char*`.


```c++
#include <iostream>
#include <string>
#include <cstring>  // for strlen, strcpy

// std::string -> const char*
std::string cppStr = "Hello from C++";
const char *cStr = cppStr.c_str();
std::cout << "As C-string: " << cStr << std::endl;
std::cout << "C-string length (strlen): " << strlen(cStr) << std::endl;

// char* -> std::string (implicit conversion via constructor)
const char *literal = "Hello from C";
std::string fromC = literal;  // constructs a std::string
std::cout << "From C literal: " << fromC << std::endl;
std::cout << "std::string length: " << fromC.length() << std::endl;

// Important: do NOT store c_str() result past the string's lifetime
// or after modifying the string — the pointer may become invalid.
// Expected output:
// As C-string: Hello from C++
// C-string length (strlen): 14
// From C literal: Hello from C
// std::string length: 12
```

## String Streams

The `<sstream>` header provides stream objects backed by strings:

- **`std::ostringstream`** — write to a string (like sprintf but safe)
- **`std::istringstream`** — read from a string (parse tokens from a string)

In C++98, there is no `std::to_string()`. The standard way to convert a number to a string is with `std::ostringstream`.


```c++
#include <iostream>
#include <string>
#include <sstream>

// C++98 way to convert number to string: ostringstream
int score = 99;
double average = 87.5;

std::ostringstream oss;
oss << "Score: " << score << ", Average: " << average;
std::string result = oss.str();

std::cout << result << std::endl;
std::cout << "Length: " << result.length() << std::endl;

// Building a string piece by piece
std::ostringstream builder;
for (int i = 1; i <= 5; i++) {
    builder << i;
    if (i < 5) builder << ", ";
}
std::cout << "List: " << builder.str() << std::endl;
// Expected output:
// Score: 99, Average: 87.5
// Length: 25
// List: 1, 2, 3, 4, 5
```

```c++
#include <iostream>
#include <string>
#include <sstream>

// Parsing a string with istringstream
std::string data = "Alice 21 3.85";
std::istringstream iss(data);

std::string name;
int age;
double gpa;

iss >> name >> age >> gpa;

std::cout << "Name: " << name << std::endl;
std::cout << "Age:  " << age  << std::endl;
std::cout << "GPA:  " << gpa  << std::endl;
// Expected output:
// Name: Alice
// Age:  21
// GPA:  3.85
```

## Modern C++ (C++11 and Beyond)

C++11 and later make string conversions and literals much more convenient.


```c++
#include <iostream>
#include <string>

// C++11: std::to_string — no more ostringstream for simple conversions
int n = 42;
double d = 3.14;
std::string s1 = std::to_string(n);
std::string s2 = std::to_string(d);
std::cout << "to_string(42): '" << s1 << "'" << std::endl;
std::cout << "to_string(3.14): '" << s2 << "'" << std::endl;

// C++11: stoi, stod — string to number
std::string numStr = "123";
std::string dblStr = "3.14";
int parsed = std::stoi(numStr);
double parsedD = std::stod(dblStr);
std::cout << "stoi('123'): " << parsed  << std::endl;
std::cout << "stod('3.14'): " << parsedD << std::endl;

// C++11: Raw string literals — no need to escape backslashes
std::string path = R"(C:\Users\alice\documents\file.txt)";
std::cout << "Path: " << path << std::endl;

std::string multiLine = R"(Line one
Line two
Line three)";
std::cout << multiLine << std::endl;
// Expected output:
// to_string(42): '42'
// to_string(3.14): '3.140000'
// stoi('123'): 123
// stod('3.14'): 3.14
// Path: C:\Users\alice\documents\file.txt
// Line one
// Line two
// Line three
```
