# Formatting and Code Conventions

Good formatting makes code readable, maintainable, and professional. When working on a team — or even alone across time — consistent style is what lets you (and others) understand code at a glance.

We establish these conventions early so that all examples throughout this course are consistent. You will be expected to follow these rules in your own code.

This notebook covers naming conventions, indentation, brace style, spacing, header/source file separation, header guards, `const` correctness, and comments.

## Naming Conventions

Names should be descriptive and follow consistent casing rules. Here are the conventions we use throughout this course:

| Category | Convention | Examples |
|---|---|---|
| Variables | camelCase | `playerScore`, `itemCount`, `firstName` |
| Functions | camelCase | `getValue()`, `printReport()`, `calculateArea()` |
| Classes | PascalCase | `MyClass`, `HttpRequest`, `LinkedList` |
| Structs | PascalCase | `Point2D`, `Config`, `PlayerData` |
| Enums | PascalCase (type), UPPER_CASE (values) | `Color { RED, GREEN, BLUE }` |
| Constants | UPPER_CASE | `MAX_SIZE`, `DEFAULT_TIMEOUT`, `PI` |
| Macros | UPPER_CASE | `ASSERT(x)`, `DEBUG_PRINT(msg)` |
| Type aliases | PascalCase | `typedef int Integer;` |

### Private members
There are two common conventions for private member variables:
- **Trailing underscore:** `value_`, `name_`, `size_`
- **m_ prefix:** `m_value`, `m_name`, `m_size`

**This course uses plain camelCase with no prefix or suffix** for member variables, since the class context makes their role clear. Pick one convention and be consistent within a project.

### What to avoid
- Single-letter names (except loop counters `i`, `j`, `k` and math variables like `x`, `y`)
- Abbreviations that are not universally understood (`numEl` is fine; `nE` is not)
- Names that differ only in case (`value` and `Value` in the same scope)
- Leading underscores (`_name`, `__impl`) — these are reserved for the compiler/standard library

````c++
#include <iostream>
#include <string>

// Constants: UPPER_CASE
const int MAX_STUDENTS = 30;
const double TAX_RATE  = 0.20;

// Enum: PascalCase type name, UPPER_CASE values
enum Direction { NORTH, SOUTH, EAST, WEST };

// Class: PascalCase
class StudentRecord {
private:
    std::string firstName;  // camelCase members (no prefix)
    std::string lastName;
    double      gpa;

public:
    // Constructor: same name as class (PascalCase)
    StudentRecord(const std::string &first, const std::string &last, double g) {
        firstName = first;
        lastName  = last;
        gpa       = g;
    }

    // Methods: camelCase
    std::string getFullName() const {
        return firstName + " " + lastName;
    }

    double getGpa() const { return gpa; }

    void printSummary() const {
        std::cout << getFullName() << " | GPA: " << gpa << std::endl;
    }
};

StudentRecord student("Alice", "Dupont", 3.8);
student.printSummary();
std::cout << "Max students: " << MAX_STUDENTS << std::endl;
// Expected output:
// Alice Dupont | GPA: 3.8
// Max students: 30
````

## Indentation and Braces

### Indentation
Use **4 spaces** per level of indentation. Do not use tabs — different editors render tabs differently, which breaks visual alignment.

Configure your editor to insert 4 spaces when you press Tab.

### Brace styles
There are two dominant brace styles in C/C++:

**Allman style** — opening brace on its own line:
```cpp
class MyClass
{
    void doSomething()
    {
        if (condition)
        {
            // ...
        }
    }
};
```

**K&R style** — opening brace on the same line (used in this course):
```cpp
class MyClass {
    void doSomething() {
        if (condition) {
            // ...
        }
    }
};
```

**This course uses K&R style.** Both are valid; what matters most is consistency within a project.

### Always use braces
Even for single-statement `if` bodies, always include braces. This prevents the classic "dangling else" bug and makes it safer to add lines later:

```cpp
// Bad (easy to introduce bugs):
if (x > 0)
    doSomething();

// Good:
if (x > 0) {
    doSomething();
}
```

````c++
#include <iostream>

// K&R brace style, 4-space indent
int fibonacci(int n) {
    if (n <= 0) {
        return 0;
    } else if (n == 1) {
        return 1;
    }
    int prev = 0;
    int curr = 1;
    for (int i = 2; i <= n; i++) {
        int next = prev + curr;
        prev = curr;
        curr = next;
    }
    return curr;
}

for (int i = 0; i <= 7; i++) {
    std::cout << "fib(" << i << ") = " << fibonacci(i) << std::endl;
}
// Expected output:
// fib(0) = 0
// fib(1) = 1
// fib(2) = 1
// fib(3) = 2
// fib(4) = 3
// fib(5) = 5
// fib(6) = 8
// fib(7) = 13
````

## Spacing Rules

Consistent spacing makes code much easier to scan.

### Good spacing:
```cpp
int total = a + b;              // spaces around binary operators
int result = (a + b) * (c - d); // spaces inside subexpressions
void print(int x, int y);       // space after comma
if (x > 0) {                    // space before opening brace
for (int i = 0; i < n; i++) {   // space after semicolons in for
```

### Bad spacing:
```cpp
int total=a+b;                  // missing spaces around operators
void print(int x,int y);        // missing space after comma
if(x>0){                        // missing space before ( and {
int arr[10] ;                   // space before semicolon (wrong)
```

### Specific rules:
- Space **before** opening brace `{` on the same line
- Space **after** commas in parameter lists and initializer lists
- Space **around** binary operators: `=`, `+`, `-`, `*`, `/`, `==`, `!=`, `<`, `>`, etc.
- **No** space between a function name and its opening parenthesis: `getValue()` not `getValue ()`
- **No** space before a semicolon
- Space **after** `if`, `for`, `while`, `switch` keywords: `if (...)` not `if(...)`
- Unary operators (`!`, `++`, `--`, `*` for dereference, `&` for address-of) hug their operand: `!flag`, `*ptr`, `++i`

````c++
#include <iostream>

// Well-spaced code
int countPositive(int arr[], int size) {
    int count = 0;
    for (int i = 0; i < size; i++) {
        if (arr[i] > 0) {
            count++;
        }
    }
    return count;
}

int numbers[] = {-3, 1, 4, -1, 5, 9, -2, 6};
int sz = sizeof(numbers) / sizeof(numbers[0]);
std::cout << "Positive numbers: " << countPositive(numbers, sz) << std::endl;
// Expected output: Positive numbers: 5
````

## Header Files vs Source Files

In C++, code is typically split across two types of files:

| File type | Extension | Contains |
|---|---|---|
| Header file | `.hpp` (C++) or `.h` (C-compatible) | Class declarations, function prototypes, type definitions, constants |
| Source file | `.cpp` | Function definitions, method implementations |

**Why separate them?**
- Headers are `#include`d by other files. If you put definitions in headers and include them in multiple `.cpp` files, you get linker errors (multiple definitions).
- Headers form the **public interface** of your code. Source files contain the **implementation details** that others don't need to see.
- Faster compilation: when you change a `.cpp` file, only that file needs recompiling.

### Example: Counter class split across files

**Counter.hpp** (declaration only):
```cpp
#ifndef COUNTER_HPP
#define COUNTER_HPP

class Counter {
private:
    int count;

public:
    Counter(int initial);
    void increment();
    void decrement();
    int  getCount() const;
};

#endif  // COUNTER_HPP
```

**Counter.cpp** (definitions):
```cpp
#include "Counter.hpp"

Counter::Counter(int initial) {
    count = initial;
}

void Counter::increment() { count++; }
void Counter::decrement() { count--; }
int  Counter::getCount() const { return count; }
```

**main.cpp** (usage):
```cpp
#include <iostream>
#include "Counter.hpp"

int main() {
    Counter c(0);
    c.increment();
    c.increment();
    std::cout << c.getCount() << std::endl;  // 2
}
```

In these notebooks we define everything in a single cell for convenience, but in real projects always split declaration from definition.

## Header Guards

If the same header is `#include`d more than once in a translation unit (directly or through other headers), the compiler will see duplicate declarations and report errors.

**Header guards** prevent this using the preprocessor:

```cpp
// Filename: MyClass.hpp
#ifndef MYCLASS_HPP   // "if not defined"
#define MYCLASS_HPP   // define the guard symbol

class MyClass {
    // ...
};

#endif  // MYCLASS_HPP  <- comment is optional but helpful
```

The first time the compiler processes this file, `MYCLASS_HPP` is not defined, so it processes everything and defines `MYCLASS_HPP`. The second time, `MYCLASS_HPP` is already defined, so everything between `#ifndef` and `#endif` is skipped.

### Naming the guard
Use the filename in UPPER_CASE with non-alphanumeric characters replaced by underscores:
- `MyClass.hpp` → `MYCLASS_HPP`
- `http_utils.hpp` → `HTTP_UTILS_HPP`
- `network/socket.hpp` → `NETWORK_SOCKET_HPP`

### Alternative: #pragma once
Many compilers support `#pragma once` as a simpler alternative:
```cpp
#pragma once

class MyClass { /* ... */ };
```
It achieves the same effect but is not part of the C++ standard (it is a widely-supported extension). **For this course, use the `#ifndef` guard form** — it is portable and standard.

## const Correctness

`const` correctness means using `const` everywhere it is appropriate — not just for constants, but also for:
- **Parameters passed by reference** that the function does not modify
- **Member functions** that do not modify the object's state
- **Return values** (less common, but sometimes useful)

Benefits:
- The compiler catches accidental modifications.
- The interface communicates intent: callers know a `const` function is safe to call on a const object.
- Enables the compiler to optimise more aggressively.

````c++
#include <iostream>
#include <string>

class BankAccount {
private:
    std::string owner;
    double      balance;

public:
    BankAccount(const std::string &ownerName, double initialBalance) {
        owner   = ownerName;
        balance = initialBalance;
    }

    // const member functions: promise not to modify *this
    std::string getOwner()  const { return owner;   }
    double      getBalance() const { return balance; }

    void printStatement() const {
        std::cout << "Account: " << owner
                  << " | Balance: $" << balance << std::endl;
    }

    // Non-const: modifies the object
    void deposit(double amount) {
        if (amount > 0) {
            balance += amount;
        }
    }

    void withdraw(double amount) {
        if (amount > 0 && amount <= balance) {
            balance -= amount;
        } else {
            std::cout << "Insufficient funds or invalid amount." << std::endl;
        }
    }
};

// Pass by const reference: we only read, do not modify
void printBalance(const BankAccount &account) {
    // account.deposit(100);  // COMPILE ERROR: cannot call non-const on const ref
    std::cout << account.getOwner() << ": $" << account.getBalance() << std::endl;
}

BankAccount alice("Alice", 1000.0);
alice.deposit(250.0);
alice.withdraw(100.0);
printBalance(alice);
// Expected output: Alice: $1150
````

## Comments

### Types of comments

```cpp
// Single-line comment — use for brief explanations on or above a line

/*
 * Multi-line comment — use for longer explanations
 * that span several lines.
 */

/**
 * Documentation comment (Doxygen style)
 * @param x the input value
 * @return the square of x
 */
int square(int x);
```

### Good vs bad comments

**Bad comment** — restates what the code obviously does:
```cpp
i++;  // increment i by 1
```

**Good comment** — explains *why*, not *what*:
```cpp
i++;  // skip the header row in the CSV
```

**Bad comment** — describes something that is clear from the name:
```cpp
// Calculate the area
double getArea() { return width * height; }
```

**Good comment** — warns about a non-obvious constraint:
```cpp
// Assumes angle is in radians, not degrees
double computeSine(double angle);
```

### Principles
- **Code should be self-documenting.** Good names and clear structure reduce the need for comments.
- Comment the *why*, not the *what*.
- Keep comments up to date — an outdated comment is worse than no comment.
- Use `//` for single-line comments (preferred in C++).
- Use `/* */` for temporarily disabling a block of code.
- Write documentation comments for every public function and class in your header files.

````c++
#include <iostream>

/**
 * Converts a temperature from Celsius to Fahrenheit.
 * Formula: F = C * 9/5 + 32
 * @param celsius temperature in degrees Celsius
 * @return temperature in degrees Fahrenheit
 */
double celsiusToFahrenheit(double celsius) {
    return celsius * 9.0 / 5.0 + 32.0;
}

// Temperatures that matter for the course project
const double BODY_TEMP_C   = 37.0;   // normal human body temperature
const double WATER_BOIL_C  = 100.0;  // sea level
const double WATER_FREEZE_C = 0.0;

std::cout << "Body temp:    " << celsiusToFahrenheit(BODY_TEMP_C)    << "F" << std::endl;
std::cout << "Boiling:      " << celsiusToFahrenheit(WATER_BOIL_C)   << "F" << std::endl;
std::cout << "Freezing:     " << celsiusToFahrenheit(WATER_FREEZE_C) << "F" << std::endl;
// Expected output:
// Body temp:    98.6F
// Boiling:      212F
// Freezing:     32F
````

## Modern C++ (C++11 and Beyond)

C++11 and later standards introduced several attributes and language features that help enforce good coding conventions at the compiler level.

### [[nodiscard]] — warn when a return value is ignored

If a function's return value is important (e.g., an error code, or a newly created resource), mark it `[[nodiscard]]` so the compiler warns if the caller ignores it:

````c++
#include <iostream>

// [[nodiscard]] added in C++17, but shown here as a preview
// The compiler will warn if the return value of this function is discarded
[[nodiscard]] int openFile(const char *path) {
    // Returns a file descriptor; -1 on error
    return -1;  // simplified
}

int fd = openFile("data.txt");  // OK: we use the return value
std::cout << "File descriptor: " << fd << std::endl;
// openFile("data.txt");        // would generate a compiler warning: discarded return value
// Expected output: File descriptor: -1
````

### override — make virtual method overriding explicit

In C++98, overriding a virtual function silently does nothing if the signature doesn't match exactly. C++11's `override` keyword makes the intention explicit and causes a compile error if the signature doesn't match:

```cpp
class Animal {
public:
    virtual void makeSound() const;
};

// C++98 — silently wrong if signature mismatches:
class Dog : public Animal {
public:
    void makeSound();  // missing const — creates a NEW function, doesn't override!
};

// C++11 — catches the error at compile time:
class Dog : public Animal {
public:
    void makeSound() const override;  // error if signature doesn't match base
};
```

### =delete — prevent unwanted operations

C++11 lets you explicitly delete copy constructors, assignment operators, or any other function you want to prohibit:

```cpp
class Singleton {
public:
    static Singleton& getInstance() { static Singleton s; return s; }

    // Prevent copying and assignment
    Singleton(const Singleton&)            = delete;
    Singleton& operator=(const Singleton&) = delete;

private:
    Singleton() {}
};
```

### [[deprecated]] — warn users of outdated APIs

```cpp
[[deprecated("Use newFunction() instead")]]
void oldFunction() { /* ... */ }
```

When someone calls `oldFunction()`, the compiler emits a warning with the message. This is far better than a comment that might be missed.
