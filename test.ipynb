# C++ Kernel Test
Testing basic C++ features in this notebook.

```c++17
#include <iostream>
std::cout << "Hello, World!" << std::endl;
```

## Variables and Types
Basic variable declarations in C++.

```c++17
int x = 42;
double pi = 3.14159;
std::string name = "C++";
std::cout << x << ", " << pi << ", " << name << std::endl;
```

## Auto Keyword
C++ can infer types with `auto`.

```c++17
auto a = 10;
auto b = 3.14;
auto c = std::string("auto!");
std::cout << a << " " << b << " " << c << std::endl;
```

## For Loop
Iterating with a range-based for loop.

```c++17
#include <vector>
std::vector<int> nums = {1, 2, 3, 4, 5};
for (auto n : nums) {
    std::cout << n << " ";
}
std::cout << std::endl;
```

## Functions
Defining and calling a simple function.

```c++17
int add(int a, int b) {
    return a + b;
}
std::cout << add(3, 7) << std::endl;
```

## Lambda
C++ lambdas — anonymous functions.

```c++17
auto square = [](int x) { return x * x; };
std::cout << square(5) << std::endl;
```

## Structs
Grouping data together with a struct.

```c++17
struct Point {
    double x, y;
};
Point p = {3.0, 4.0};
std::cout << "Point: (" << p.x << ", " << p.y << ")" << std::endl;
```

## String Operations
Common string manipulations.

```c++17
std::string s = "Hello, World!";
std::cout << "Length: " << s.length() << std::endl;
std::cout << "Substr: " << s.substr(0, 5) << std::endl;
std::cout << "Upper H at: " << s.find('W') << std::endl;
```

## Map
Key-value storage with `std::map`.

```c++17
#include <map>
std::map<std::string, int> scores;
scores["Alice"] = 95;
scores["Bob"] = 87;
for (auto& [name, score] : scores) {
    std::cout << name << ": " << score << std::endl;
}
```
